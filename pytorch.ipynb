```python
print("=== Tensor Operations ===")

a = torch.tensor([1.0, 2.0, 3.0])
b = torch.tensor([4.0, 5.0, 6.0])

# Element-wise operations
print("Addition:", a + b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Power:", a ** 2)

# Matrix operations
matrix_a = torch.randn(3, 4)
matrix_b = torch.randn(4, 2)
result = torch.mm(matrix_a, matrix_b)  # Matrix multiplication
print("\nMatrix multiplication shape:", result.shape)

# Reshaping
original = torch.randn(2, 6)
reshaped = original.view(3, 4)  # Same data, different shape
print("\nOriginal shape:", original.shape)
print("Reshaped:", reshaped.shape)
```

```python
print("=== Autograd Demo ===")

# Create a tensor that tracks operations for gradients
x = torch.tensor([2.0], requires_grad=True)
print("x:", x)

# Do some operations
y = x ** 2 + 3 * x + 1
print("y = x² + 3x + 1 =", y.item())

# Compute gradients automatically!
y.backward()  # This calculates dy/dx

print("Gradient (dy/dx):", x.grad.item())
# Manual calculation: dy/dx = 2x + 3 = 2(2) + 3 = 7 ✓

# More complex example
print("\n=== Multi-variable Example ===")
a = torch.tensor([3.0], requires_grad=True)
b = torch.tensor([4.0], requires_grad=True)

z = a ** 2 + b ** 2  # z = a² + b²
z.backward()

print("dz/da:", a.grad.item())  # Should be 2a = 6
print("dz/db:", b.grad.item())  # Should be 2b = 8
```

```python
import torch.nn as nn

print("=== Building a Neural Network ===")

# Define a simple network
class MyFirstNetwork(nn.Module):
    def __init__(self):
        super().__init__()
        # Define layers
        self.layer1 = nn.Linear(10, 5)  # 10 inputs -> 5 outputs
        self.layer2 = nn.Linear(5, 2)   # 5 inputs -> 2 outputs
        self.relu = nn.ReLU()           # Activation function

    def forward(self, x):
        # Define how data flows through the network
        x = self.layer1(x)
        x = self.relu(x)
        x = self.layer2(x)
        return x

# Create the network
model = MyFirstNetwork()
print(model)

# Test with random input
test_input = torch.randn(1, 10)  # Batch of 1, 10 features
output = model(test_input)
print("\nInput shape:", test_input.shape)
print("Output shape:", output.shape)
print("Output:", output)
```
